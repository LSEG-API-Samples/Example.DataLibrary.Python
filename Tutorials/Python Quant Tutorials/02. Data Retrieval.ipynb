## `ld.content.symbol_conversion`

The function `ld.content.symbol_conversion` allows you to **convert symbols** from one symbology type to another.

```python
help(ld.content.symbol_conversion)
```

Assume the following set of `RICs`.

```python
rics = [
    'AAPL.O',  # Apple stock
    'AMZN.O',  # Amazon stock
    'TSLA.O',  # Tesla stock
    '.SPX',  # S&P 500 stock index
]
```

The **conversion** works as follows:

```python
isin = ld.content.symbol_conversion.Definition(
    symbols=rics,
    from_symbol_type=ld.content.symbol_conversion.SymbolTypes.RIC,
    to_symbol_types=[ld.content.symbol_conversion.SymbolTypes.ISIN]
    ).get_data().data.df
```

```python
isin
```

And also the other way round, after a conversion of data structures:

```python
isins = list(isin['IssueISIN'].dropna().values)
```

```python
isins
```

```python
isin = ld.content.symbol_conversion.Definition(
    symbols=isins,
    from_symbol_type=ld.content.symbol_conversion.SymbolTypes.ISIN,
    to_symbol_types=[ld.content.symbol_conversion.SymbolTypes.RIC]
    ).get_data().data.df
```

## `ld.get_data()`

The function `ld.get_data()` is a **general purpose data retrieval** function.

```python
help(ld.get_data)
```

```python
data_grid = ld.get_data(
    ['IBM', 'GOOG.O', 'MSFT.O'],
    ['TR.TotalReturnYTD', 'TR.WACCBeta'])
```

```python
data_grid
```

```python
data_grid.set_index('Instrument')[['YTD Total Return', 'Beta']].iplot(
    kind='bar', subplots=True)
```

## `ld.get_history()`

The function `ld.get_history()` retrieves **time series data**.

```python
help(ld.get_history)
```

For instance, the following code retries **daily closing prices**  for the four `RICs` as defined above.

```python
rics
```

```python
data = ld.get_history(
    rics,  # the RICs
    fields='TR.PriceClose',  # close field
    start='2024-01-01',  # start date
    end='2024-05-20',  # end date
    interval='1D').dropna()
```

```python
data.head()
```

```python
data.info()
```

Making use of `Cufflinks`, the data is visualized below.

```python
data.normalize().iplot()
```

## `ld.news.get_headlines`

The function `ld.news.get_headlines` allows you to search for and retrieve **news headlines**, including `storyId` values needed to retrieve the full news text.

```python
help(ld.news.get_headlines)
```

A `query` string might contain `RICs` and other words to be searched for.

```python
rics[2]
```

```python
news = ld.news.get_headlines(
    'R:%s PRODUCTION IN ENGLISH' % rics[2],
    start='2024-05-01',
    end='2024-05-14',
    count=10)
```

```python
news
```

### Story

The function `ld.content.news.story.Definition` retrieves the full text of a **news stroy** given the `storyId` value.

```python
help(ld.content.news.story.Definition)
```

The `storyId` values are stored in the respective column of the `news` `DataFrame` object as created above.

```python
news['storyId']
```

To read a news story, **pick out one story via its storyId and display the full text** provided as HTML code.

```python
storyId = news['storyId'][0]
```

```python
from IPython.display import HTML
```

```python
response = ld.content.news.story.Definition(storyId).get_data()
print(response.data.story.title, '\n')
HTML(response.data.story.content.html)
```
