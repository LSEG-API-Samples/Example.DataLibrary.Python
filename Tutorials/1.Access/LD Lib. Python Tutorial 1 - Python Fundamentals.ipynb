### If Else Statements
If Else statements are very simple and extremely powerful:

```python
if a == 44.9:
    print("a is an float")
```

```python
if a >= 44:
    print("a is a large number")
```

```python
if a < 44:
    print("a is a small number number")
```

```python
if a != 44.9:
    print("a is *not* 44.9")
else:
    print("a is 44.9")
```

### For Loops

You can iterate through logic in Python easily:

```python
factorial = 1
for i in [2, 3, 4, 5]:
    factorial = factorial * i
    print(factorial)
```

#### List Comprehensions

A faster way to loop through data is to 

```python
[i+10 for i in range(10)] # Note that `range` starts at 0 is exclusive with respect to its upper limit
```

#### Lambda

A faster way to create functions is using Lambda Functions, e.g.:

```python
add = lambda a, b : a * b
```

```python
add(1, 100)
```

```python
[add(i, 100) for i in range(10)]
```

```python
[(lambda x: x**2)(i) for i in range(10)] # lambda function inside a list comprehention
```

#### Zips

```python
for i, j in zip([1, 2, 3], [-1, -2, -3]):
    print(f"i in {i}, and j is {j}.")
```
