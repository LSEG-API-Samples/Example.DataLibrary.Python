There are two sessions you can use to authenticate yourself to the LSEG Data Platform, the Desktop Session and Platform Session.

The Desktop Session allows you to access data in LSEG's Data Platform via the Desktop Workspace Application on the machine where you are running your code, meaning taht you need to have Workspace running on the machine where you are running your Python code. #### M&A League Tables

One of the great things about the CODEBOOK app is that you can reuse the screen criteria from the Screener APP. In this example we screened for M&A Deals YTD globally

```python
# use SCREENER to produce list of deals meeting League Table criteria
lt_screen = "SCREEN(U(IN(DEALS)/*UNV:DEALSMNA*/),  relativedate(TR.MnARankDate,LY)/*dt:Date*/, \
            TR.MnAIsEligForStdRanking==true,TR.MnAHasDebtTender = false,TR.MnAHasLoanModify = false, \
            TR.MnAIsExchangeOffer = false,TR.MnAIsOPMAndExclFromRanking = false,TR.MnAIsEquityCarveout = false, CURN=USD)"
# define list of fields and then using get_data from Eikon Data API create a dataframe with the screen and fields
lt_fields = [
    'TR.MnALeagueAdvisor(DealMnALTType=FINADV, DealMnAIsParent=true,LTAllocType=BF, ExclIneligible=true)',
    'TR.MnARankValueIncNetDebt(Scale=6,Curn=USD)']
# Send data request and package responce
lt = ld.get_data(lt_screen,lt_fields)
lt
```

Quick Tip: If you need help defining your field selection use CODECREATOR to simply create the lines of code you need and click on 'Open in CODEBOOK'.

```python
# refine dataframe to meet specific criteria for analysis. 
lt2 = lt[lt['M&A League Advisor'] != 'None'] #creates a new table removing results showing None
lt3 = lt2[lt2['M&A League Advisor'] != 'No Information'] #takes that table and removes results showing No Info
lt3 = lt3.dropna() #drops any NaNs that might be left so we have a clean data set
lt3_filtered = lt3[lt3['Rank Value inc. Net Debt of Target'] > 0] # filter out deals that aren't rank valued so we can compare properly
lt3_filtered
```

```python
# group all deals by League Advisor so we can sum up their Rank Values
lt3_filtered = lt3_filtered.groupby('M&A League Advisor')['Rank Value inc. Net Debt of Target'].sum()[0:]
lt4 = pd.DataFrame(lt3_filtered)
```

```python
# Create a table with the 10 largest M&A League Advisors based on their total Rank Value 
lt5 = lt4.nlargest(10,'Rank Value inc. Net Debt of Target')
lt5
```

```python
import plotly.express as px # the plotly libraries for visualization and UI
import plotly.graph_objects as go

# create pie chart using Plotly by using the above dataframe and apply styling and formatting
fig2 = px.pie(lt5, values='Rank Value inc. Net Debt of Target', title='M&A League Table', names=lt5.index)
fig2.layout.template = 'plotly_dark'
fig2.update_traces(textposition='inside')
fig2.update_layout(uniformtext_minsize=12, uniformtext_mode='hide')
fig2.update(layout=dict(title=dict(x=0.43)))
fig2.show()
```

#### IPOs

LSEG provides industry-leading equity and equity-related transaction information and league tables to the global deal-making industry, comprising over 300,000 global equity and equity-related deals since the 1970s. LSEG relies upon direct deal submissions from global deal makers coupled with rigorous sourcing of regulatory filings, newswires, and company press releases to gather timely and detailed equity issuance information.

```python
# Once again use SCREENER to pull list of IPOs meeting our criteria YTD
ipo_screen = "SCREEN(U(IN(DEALS)/*UNV:DEALSEQ*/), TR.NIisECM=True, \
            TR.NIIsRankEligible==true, relativedate(TR.NIIssueDate,YTD)/*dt:Date*/, \
            TR.NIIsIpo==true, IN(TR.NITransactionStatus,'LIVE'), CURN=USD)"
# input ipo screen results as instrument list and pull ipo attributes via fields from Eikon Data API. 
ipos = ld.get_data(
    ipo_screen,['TR.NISdcPackageNumber',
                'TR.NIIsIpo',
                'TR.NIIssueDate',
                'TR.NIIssuer',
                'TR.NITRBCIndustry',
                'TR.NIIssuerPermId',
                'TR.NIIssuerNation',
                'TR.NIRegionHQ',
                'TR.NIOfferPrice',
                'TR.NIProceedsAmtInclOverallotSoldAllMkts(Scale=6)',
                'TR.NIIssuerExchTicker',
                'TR.NISecurityTypeAllMkt',
                'TR.NIBookOrCoManager',
                'TR.NIHasLockup',
                'TR.NILockupDays',
                'TR.NILockupShares'])
ipos.head()
```

```python
# remove rows with duplicate package identifiers which support the same deal and cause double count
ipos = ipos[ipos['Package Identifier'].replace('', pd.NA).notna()]
ipos.describe()
```

```python
industry = 'New Issues TRBC Industry'
fig3 = go.Figure()
trace = go.Bar(x=ipos[industry].value_counts().head(15).values, 
               y=ipos[industry].value_counts().head(15).index,
               orientation='h')
fig3.add_trace(trace)
fig3.layout.template = 'plotly_dark'
fig3
```

#### Dataquery

Leonid's article "[Dataquery – a syntax builder for screening with Eikon Data API](https://developers.lseg.com/en/article-catalog/article/dataquery-a-syntax-builder-for-screening-with-eikon-data-api)" is also great. In that article, he shows how one can use his [dataquery.py](https://github.com/LSEG-API-Samples/dataquery/blob/master/dataquery.py) document to make screens from scratch.

```python
from dataquery import *

screen = SCREEN.express.universe(
    Equity(active=True, public=True, primary=True)).conditions(
        IN('TR.ImmediateParentCountryHQ', "United States"),
        # IN('TR.CoRTradingCountryCode', 'US'),
        TOP(field='TR.CompanyMarketCap', bound=10, metric=nnumber)
        ).currency('USD').query

screen
```

```python
df = ld.get_data(
    universe='SCREEN(U(IN(Equity(primary,public,active))),IN(TR.ImmediateParentCountryHQ,"United States"),TOP(TR.CompanyMarketCap,10,nnumber),CURN=USD)',
    fields=['TR.CommonName','TR.CompanyMarketCap(Scale=6)','TR.ExchangeName',
            'TR.TRBCBusinessSector','TR.TotalReturn3Mo'])

display(df)
```

## Platform Session

Platform sessions allow us to use the LD Lib. without the Workspace Desktop App open. It was created for servers, and it is still aimed for data hungry users who require little human interation with the code at hand (e.g.: overnight data retraival). With that said, usecases have grown in breadth and complexity thoughout the years, and people may want such 'ping' services (i.e.: the ability to 'ping', to directly contact, the services that deliver data/calculations, without going through the Desktop App Proxy).

Platform session use-cases are a little more complicated to the Desktop session ones. We can authenticate ourselves with the Workspace Desktop App open with simply just the Python line:

```python
import lseg.data as ld
ld.open_session(
    config_name="C:/Example.DataLibrary.Python-main/Configuration/lseg-data.config.json",
    name="platform.ldp")
```

```python
rics = ld.discovery.Screener(
    'U(IN(Equity(active,public,primary))/*UNV:Public*/), ' +
    'TR.CompanyMarketCap(Scale=6)>=5000, IN(TR.ExchangeMarketIdCode,"XNYS"), ' +
    'IN(TR.TRBCBusinessSectorCode,"5010","5020","5030"), TR.TotalReturn3Mo>=15, ' +
    'CURN=USD')

df = ld.get_data(
    universe=rics,
    fields=['TR.CommonName','TR.CompanyMarketCap(Scale=6)','TR.ExchangeName',
            'TR.TRBCBusinessSector','TR.TotalReturn3Mo'])
 
display(df)
```

```python
ld.close_session()
```
