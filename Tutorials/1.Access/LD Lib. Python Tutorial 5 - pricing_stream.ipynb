#### Pricing Streams

```python
df = ld.get_data(
    universe=['JPY='],
    fields=['BID','ASK'])
df.index = [datetime.datetime.now().time()]
df.index.name = 'Time'
df
```

```python
def display_data(data, instrument, stream):
    global df
    current_time = datetime.datetime.now().time()
    # print(current_time, "- Data received for", instrument)
    data.index = [current_time]
    df = pd.concat([df, data], ignore_index=True)
    df.index.name = 'Time'
    fig = px.line(df[['BID','ASK']])
    fig.update_layout(title='JPY=', template='plotly_dark')
    clear_output(wait=True)
    fig.show()
```

```python
stream = ld.open_pricing_stream(
    universe=['JPY='],
    fields=['BID','ASK'],
    on_data=display_data)
```

```python
stream.close()
```

#### Pricing Stream Recorder

```python
from IPython.display import display, clear_output
```

```python
def display_data(data, instrument, stream):
    clear_output(wait=True)
    current_time = datetime.datetime.now().time()
    print(current_time, "- Data received for", instrument)
    display(data)
```

```python
stream = ld.open_pricing_stream(
    universe=['GBP=', 'EUR=', 'JPY=', "BTC="],
    fields=['BID', 'ASK'],
    on_data=display_data)
```

```python
# help(stream.recorder.record)
```

```python
def callback(dataframe, recorder):
    print(dataframe)

stream.recorder.record(
    frequency='tick',
    duration="15s",
    on_data=callback
    )
```

```python
stream.recorder.stop()
tick_history = stream.recorder.get_history()
display(tick_history)
```

```python
stream.close()
```

## Platform Session

Platform sessions allow us to use the LD Lib. without the Workspace Desktop App open. It was created for servers, and it is still aimed for data hungry users who require little human interation with the code at hand (e.g.: overnight data retraival). With that said, usecases have grown in breadth and complexity thoughout the years, and people may want such 'ping' services (i.e.: the ability to 'ping', to directly contact, the services that deliver data/calculations, without going through the Desktop App Proxy).

Platform session use-cases are a little more complicated to the Desktop session ones. We can authenticate ourselves with the Workspace Desktop App open with simply just the Python line:
